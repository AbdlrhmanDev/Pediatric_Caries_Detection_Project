## Annotation audit and duplicate checks

### Step 19: Inspect JSON annotations

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
import json
from collections import Counter, defaultdict
from pathlib import Path

detection_train = (
    base /
    "Pediatric dental disease detection dataset" /
    "Train"
)

detection_test = (
    base /
    "Pediatric dental disease detection dataset" /
    "Test"
)

def audit_detection_labels(folder, split_name):

    json_files = sorted(folder.rglob("*.json"))

    label_counts = Counter()
    shape_counts = Counter()
    files_per_label = defaultdict(set)

    for json_file in json_files:

        with open(json_file, "r", encoding="utf-8") as f:
            data = json.load(f)

        image_name = data.get("imagePath", json_file.stem)

        for shape in data.get("shapes", []):

            label = str(shape.get("label", "")).strip()
            shape_type = shape.get("shape_type", "unknown")

            label_counts[label] += 1
            shape_counts[shape_type] += 1
            files_per_label[label].add(image_name)

    print("\n" + "="*80)
    print(split_name)
    print("="*80)

    print("Number of JSON files:", len(json_files))

    print("\nLabels:")
    for label, count in label_counts.most_common():
        print(
            f"{label:20s} | lesions: {count:4d} | "
            f"images containing label: {len(files_per_label[label]):3d}"
        )

    print("\nShape types:")
    for shape_type, count in shape_counts.items():
        print(shape_type, ":", count)

    return label_counts, files_per_label


train_counts, train_files = audit_detection_labels(
    detection_train,
    "TRAIN"
)

test_counts, test_files = audit_detection_labels(
    detection_test,
    "TEST"
)
```

### Step 20: Run annotation audit and duplicate checks step

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
all_counts = train_counts + test_counts

all_labels = sorted(
    set(train_files.keys()) | set(test_files.keys())
)

print("="*80)
print("ENTIRE DETECTION DATASET")
print("="*80)

for label in all_counts.most_common():

    label_name = label[0]
    lesion_count = label[1]

    images = set()

    if label_name in train_files:
        images.update(train_files[label_name])

    if label_name in test_files:
        images.update(test_files[label_name])

    print(
        f"{label_name:20s} | "
        f"total lesions: {lesion_count:4d} | "
        f"images: {len(images):3d}"
    )
```

### Step 28: Check exact duplicate images

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
import hashlib
from pathlib import Path

detection_images = list(
    (base / "Pediatric dental disease detection dataset").rglob("*.png")
)

supp_images = list(
    (
        base /
        "Childrens dental caries segmentation dataset" /
        "Supplemental content93" /
        "images"
    ).glob("*.png")
)

print("Detection images:", len(detection_images))
print("Supplemental images:", len(supp_images))


def md5_hash(path):
    h = hashlib.md5()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            h.update(chunk)

    return h.hexdigest()


detection_hashes = {
    md5_hash(p): p
    for p in detection_images
}

supp_hashes = {
    md5_hash(p): p
    for p in supp_images
}

duplicates = set(detection_hashes) & set(supp_hashes)

print("\nExact duplicate images found:", len(duplicates))

for h in list(duplicates)[:20]:
    print(
        detection_hashes[h].name,
        "<->",
        supp_hashes[h].name
    )
```

### Step 40: Inspect JSON annotations

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
import json
from sklearn.model_selection import train_test_split

detection_base = (
    base /
    "Pediatric dental disease detection dataset"
)

original_train = detection_base / "Train"
original_test = detection_base / "Test"


def collect_cases(folder):

    cases = []

    for json_file in sorted(folder.rglob("*.json")):

        with open(json_file, "r", encoding="utf-8") as f:
            data = json.load(f)

        has_caries = any(
            str(shape.get("label", "")).strip() == "龋病"
            for shape in data.get("shapes", [])
        )

        cases.append({
            "json": json_file,
            "has_caries": has_caries
        })

    return cases


development_cases = collect_cases(original_train)

print("Development images:", len(development_cases))
print(
    "Caries-positive:",
    sum(c["has_caries"] for c in development_cases)
)
print(
    "Caries-negative:",
    sum(not c["has_caries"] for c in development_cases)
)
```

### Step 41: Create the development split

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
train_cases, val_cases = train_test_split(
    development_cases,
    test_size=0.20,
    random_state=42,
    stratify=[
        c["has_caries"]
        for c in development_cases
    ]
)

print("Training:", len(train_cases))
print("Validation:", len(val_cases))

print(
    "Train positive:",
    sum(c["has_caries"] for c in train_cases)
)

print(
    "Validation positive:",
    sum(c["has_caries"] for c in val_cases)
)
```

```python
import shutil
from PIL import Image


CARIES_LABEL = "龋病"
CLASS_ID = 0


def find_image(json_file, data):

    image_name = Path(data["imagePath"]).name

    # البحث داخل نفس الـsplit
    candidates = list(
        json_file.parent.parent.rglob(image_name)
    )

    if candidates:
        return candidates[0]

    # بديل بالـstem
    for ext in [".png", ".jpg", ".jpeg", ".bmp"]:
        candidates = list(
            json_file.parent.parent.rglob(
                Path(image_name).stem + ext
            )
        )

        if candidates:
            return candidates[0]

    return None


def convert_case(case, split):

    json_file = case["json"]

    with open(json_file, "r", encoding="utf-8") as f:
        data = json.load(f)

    image_path = find_image(json_file, data)

    if image_path is None:
        print("❌ Image not found:", json_file)
        return 0

    image_width = data.get("imageWidth")
    image_height = data.get("imageHeight")

    if not image_width or not image_height:

        with Image.open(image_path) as img:
            image_width, image_height = img.size

    # نسخ الصورة
    destination_image = (
        YOLO_DIR /
        "images" /
        split /
        image_path.name
    )

    shutil.copy2(
        image_path,
        destination_image
    )

    yolo_lines = []

    for shape in data.get("shapes", []):

        label = str(
            shape.get("label", "")
        ).strip()

        if label != CARIES_LABEL:
            continue

        points = shape.get("points", [])

        if len(points) < 2:
            continue

        x1 = min(points[0][0], points[1][0])
        y1 = min(points[0][1], points[1][1])

        x2 = max(points[0][0], points[1][0])
        y2 = max(points[0][1], points[1][1])

        box_width = x2 - x1
        box_height = y2 - y1

        x_center = (x1 + x2) / 2
        y_center = (y1 + y2) / 2

        # Normalize 0–1
        x_center /= image_width
        y_center /= image_height

        box_width /= image_width
        box_height /= image_height

        # حماية إضافية
        x_center = max(0, min(1, x_center))
        y_center = max(0, min(1, y_center))

        box_width = max(0, min(1, box_width))
        box_height = max(0, min(1, box_height))

        yolo_lines.append(
            f"{CLASS_ID} "
            f"{x_center:.6f} "
            f"{y_center:.6f} "
            f"{box_width:.6f} "
            f"{box_height:.6f}"
        )

    label_path = (
        YOLO_DIR /
        "labels" /
        split /
        f"{image_path.stem}.txt"
    )

    with open(label_path, "w") as f:
        f.write("\n".join(yolo_lines))

    return len(yolo_lines)
```

### Step 69: Define reusable analysis functions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
import numpy as np
from pathlib import Path
from PIL import Image


def yolo_gt_to_xyxy(label_path, W, H):

    boxes = []

    if not label_path.exists():
        return boxes

    content = label_path.read_text().strip()

    if not content:
        return boxes

    for line in content.splitlines():

        cls, xc, yc, bw, bh = map(float, line.split())

        xc *= W
        yc *= H
        bw *= W
        bh *= H

        x1 = xc - bw / 2
        y1 = yc - bh / 2
        x2 = xc + bw / 2
        y2 = yc + bh / 2

        boxes.append(
            [x1, y1, x2, y2]
        )

    return boxes


def calculate_iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])

    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = max(0, x2 - x1) * max(0, y2 - y1)

    area1 = (
        (box1[2] - box1[0]) *
        (box1[3] - box1[1])
    )

    area2 = (
        (box2[2] - box2[0]) *
        (box2[3] - box2[1])
    )

    union = area1 + area2 - intersection

    if union <= 0:
        return 0

    return intersection / union
```

### Step 70: Generate model predictions

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
IOU_THRESHOLD = 0.50
CONF_THRESHOLD = 0.25

for image_path in six_images:

    img = Image.open(image_path)
    W, H = img.size

    label_path = (
        YOLO_DIR /
        "labels" /
        "test" /
        f"{image_path.stem}.txt"
    )

    gt_boxes = yolo_gt_to_xyxy(
        label_path,
        W,
        H
    )

    prediction = best_model.predict(
        source=str(image_path),
        imgsz=1024,
        conf=CONF_THRESHOLD,
        iou=0.50,
        device=0,
        verbose=False
    )[0]

    pred_boxes = []

    if prediction.boxes is not None:

        for box in prediction.boxes:

            coords = (
                box.xyxy[0]
                .cpu()
                .numpy()
                .tolist()
            )

            pred_boxes.append(coords)

    matched_gt = set()

    TP = 0
    FP = 0

    for pred_box in pred_boxes:

        best_iou = 0
        best_gt_index = None

        for gt_index, gt_box in enumerate(gt_boxes):

            if gt_index in matched_gt:
                continue

            iou = calculate_iou(
                pred_box,
                gt_box
            )

            if iou > best_iou:

                best_iou = iou
                best_gt_index = gt_index

        if (
            best_iou >= IOU_THRESHOLD
            and best_gt_index is not None
        ):

            TP += 1
            matched_gt.add(best_gt_index)

        else:
            FP += 1

    FN = len(gt_boxes) - len(matched_gt)

    print("\n", image_path.name)
    print("GT boxes :", len(gt_boxes))
    print("Predicted:", len(pred_boxes))
    print("TP:", TP)
    print("FP:", FP)
    print("FN:", FN)
```

```python
import pandas as pd
from PIL import Image

val_images = sorted([
    p for p in (YOLO_DIR / "images" / "val").iterdir()
    if p.suffix.lower() in [".png", ".jpg", ".jpeg", ".bmp"]
])

thresholds = [
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.40,
    0.50
]

IOU_THRESHOLD = 0.50

threshold_results = []

for conf_threshold in thresholds:

    total_TP = 0
    total_FP = 0
    total_FN = 0

    for image_path in val_images:

        img = Image.open(image_path)
        W, H = img.size

        # Ground Truth
        label_path = (
            YOLO_DIR /
            "labels" /
            "val" /
            f"{image_path.stem}.txt"
        )

        gt_boxes = yolo_gt_to_xyxy(
            label_path,
            W,
            H
        )

        # Predictions
        prediction = best_model.predict(
            source=str(image_path),
            imgsz=1024,
            conf=conf_threshold,
            iou=0.50,
            device=0,
            verbose=False
        )[0]

        pred_boxes = []

        if prediction.boxes is not None:

            for box in prediction.boxes:

                coords = (
                    box.xyxy[0]
                    .cpu()
                    .numpy()
                    .tolist()
                )

                pred_boxes.append(coords)

        matched_gt = set()

        TP = 0
        FP = 0

        for pred_box in pred_boxes:

            best_iou = 0
            best_gt_index = None

            for gt_index, gt_box in enumerate(gt_boxes):

                if gt_index in matched_gt:
                    continue

                iou = calculate_iou(
                    pred_box,
                    gt_box
                )

                if iou > best_iou:
                    best_iou = iou
                    best_gt_index = gt_index

            if (
                best_iou >= IOU_THRESHOLD
                and best_gt_index is not None
            ):
                TP += 1
                matched_gt.add(best_gt_index)

            else:
                FP += 1

        FN = len(gt_boxes) - len(matched_gt)

        total_TP += TP
        total_FP += FP
        total_FN += FN

    precision = (
        total_TP / (total_TP + total_FP)
        if (total_TP + total_FP) > 0
        else 0
    )

    recall = (
        total_TP / (total_TP + total_FN)
        if (total_TP + total_FN) > 0
        else 0
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if (precision + recall) > 0
        else 0
    )

    threshold_results.append({
        "Confidence": conf_threshold,
        "TP": total_TP,
        "FP": total_FP,
        "FN": total_FN,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    })


threshold_df = pd.DataFrame(threshold_results)

print(threshold_df.to_string(index=False))
```

### Step 147: Define the Torchvision dataset

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
class CariesDetectionDataset(Dataset):

    def __init__(self, image_dir, label_dir):

        self.image_dir = Path(image_dir)
        self.label_dir = Path(label_dir)

        self.images = sorted([
            p for p in self.image_dir.iterdir()
            if p.suffix.lower() in [".png", ".jpg", ".jpeg", ".bmp"]
        ])

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):

        image_path = self.images[idx]

        image = Image.open(image_path).convert("RGB")

        W, H = image.size

        image_tensor = torchvision.transforms.functional.to_tensor(image)

        label_path = (
            self.label_dir
            / f"{image_path.stem}.txt"
        )

        boxes = []
        labels = []

        if label_path.exists():

            content = label_path.read_text().strip()

            if content:

                for line in content.splitlines():

                    cls, xc, yc, bw, bh = map(float, line.split())

                    xc *= W
                    yc *= H
                    bw *= W
                    bh *= H

                    x1 = xc - bw / 2
                    y1 = yc - bh / 2
                    x2 = xc + bw / 2
                    y2 = yc + bh / 2

                    boxes.append([
                        x1, y1, x2, y2
                    ])

                    # Faster R-CNN:
                    # class 0 = background
                    # caries must be class 1
                    labels.append(1)

        if len(boxes) > 0:

            boxes = torch.tensor(
                boxes,
                dtype=torch.float32
            )

            labels = torch.tensor(
                labels,
                dtype=torch.int64
            )

        else:

            boxes = torch.zeros(
                (0, 4),
                dtype=torch.float32
            )

            labels = torch.zeros(
                (0,),
                dtype=torch.int64
            )

        target = {
            "boxes": boxes,
            "labels": labels,
            "image_id": torch.tensor([idx])
        }

        return image_tensor, target
```

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
params = [
    p for p in model.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    params,
    lr=1e-4,
    weight_decay=1e-4
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=3
)
```

### Step 152: Train the detection model

Trains the selected detector with the parameters shown below. Run this only after the data checks in the preceding cells pass.


```python
from copy import deepcopy

max_epochs = 40
early_stop_patience = 8

best_val_loss = float("inf")
best_state = None
epochs_without_improvement = 0

history = []

for epoch in range(max_epochs):

    # ============================
    # TRAIN
    # ============================

    model.train()

    train_loss_total = 0.0

    for images, targets in train_loader:

        images = [
            img.to(device)
            for img in images
        ]

        targets = [
            {
                k: v.to(device)
                for k, v in t.items()
            }
            for t in targets
        ]

        loss_dict = model(
            images,
            targets
        )

        losses = sum(
            loss for loss in loss_dict.values()
        )

        optimizer.zero_grad()

        losses.backward()

        optimizer.step()

        train_loss_total += losses.item()


    avg_train_loss = (
        train_loss_total /
        len(train_loader)
    )


    # ============================
    # VALIDATION LOSS
    # ============================

    model.train()

    val_loss_total = 0.0

    with torch.no_grad():

        for images, targets in val_loader:

            images = [
                img.to(device)
                for img in images
            ]

            targets = [
                {
                    k: v.to(device)
                    for k, v in t.items()
                }
                for t in targets
            ]

            loss_dict = model(
                images,
                targets
            )

            losses = sum(
                loss for loss in loss_dict.values()
            )

            val_loss_total += losses.item()


    avg_val_loss = (
        val_loss_total /
        len(val_loader)
    )

    scheduler.step(
        avg_val_loss
    )

    history.append({
        "epoch": epoch + 1,
        "train_loss": avg_train_loss,
        "val_loss": avg_val_loss
    })


    print(
        f"Epoch {epoch+1:02d}/{max_epochs} | "
        f"Train Loss: {avg_train_loss:.4f} | "
        f"Val Loss: {avg_val_loss:.4f}"
    )


    # ============================
    # EARLY STOPPING
    # ============================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss

        best_state = deepcopy(
            model.state_dict()
        )

        epochs_without_improvement = 0

        print("✅ New best model")

    else:

        epochs_without_improvement += 1

        print(
            "No improvement:",
            epochs_without_improvement
        )


    if epochs_without_improvement >= early_stop_patience:

        print("🛑 Early stopping")

        break
```

```python
CONF_THRESHOLD = 0.25
IOU_THRESHOLD = 0.50
```

### Step 159: Define reusable analysis functions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
def calculate_iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])

    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = (
        max(0, x2 - x1)
        *
        max(0, y2 - y1)
    )

    area1 = (
        (box1[2] - box1[0])
        *
        (box1[3] - box1[1])
    )

    area2 = (
        (box2[2] - box2[0])
        *
        (box2[3] - box2[1])
    )

    union = (
        area1 +
        area2 -
        intersection
    )

    if union <= 0:
        return 0.0

    return intersection / union
```

### Step 160: Keep predictions >= 0.25

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
total_TP = 0
total_FP = 0
total_FN = 0

frcnn_best.eval()

with torch.no_grad():

    for images, targets in val_loader:

        images_gpu = [
            image.to(device)
            for image in images
        ]

        outputs = frcnn_best(
            images_gpu
        )

        for output, target in zip(
            outputs,
            targets
        ):

            scores = (
                output["scores"]
                .detach()
                .cpu()
                .numpy()
            )

            boxes = (
                output["boxes"]
                .detach()
                .cpu()
                .numpy()
            )

            # Keep predictions >= 0.25
            keep = (
                scores >= CONF_THRESHOLD
            )

            pred_boxes = boxes[keep]

            gt_boxes = (
                target["boxes"]
                .detach()
                .cpu()
                .numpy()
            )

            matched_gt = set()

            TP = 0
            FP = 0

            for pred_box in pred_boxes:

                best_iou = 0
                best_gt_index = None

                for gt_index, gt_box in enumerate(
                    gt_boxes
                ):

                    if gt_index in matched_gt:
                        continue

                    iou = calculate_iou(
                        pred_box,
                        gt_box
                    )

                    if iou > best_iou:

                        best_iou = iou
                        best_gt_index = gt_index

                if (
                    best_iou >= IOU_THRESHOLD
                    and
                    best_gt_index is not None
                ):

                    TP += 1
                    matched_gt.add(
                        best_gt_index
                    )

                else:
                    FP += 1

            FN = (
                len(gt_boxes)
                -
                len(matched_gt)
            )

            total_TP += TP
            total_FP += FP
            total_FN += FN
```

### Step 161: Run faster r-cnn cross-validation step

Displays a checkpoint value so the current notebook state can be verified before continuing.


```python
precision = (
    total_TP /
    (total_TP + total_FP)
    if (total_TP + total_FP) > 0
    else 0
)

recall = (
    total_TP /
    (total_TP + total_FN)
    if (total_TP + total_FN) > 0
    else 0
)

f1 = (
    2 * precision * recall /
    (precision + recall)
    if (precision + recall) > 0
    else 0
)

print("=" * 60)
print("FASTER R-CNN — FOLD 1 DETECTION RESULTS")
print("=" * 60)

print("TP        :", total_TP)
print("FP        :", total_FP)
print("FN        :", total_FN)

print("Precision :", precision)
print("Recall    :", recall)
print("F1        :", f1)
```

```python
for fold in [2, 3, 4, 5]:

    print("\n" + "=" * 80)
    print(f"FASTER R-CNN — STARTING FOLD {fold}")
    print("=" * 80)

    # =========================================================
    # DATASET
    # =========================================================

    FOLD_DIR = (
        PROJECT_DIR
        / "cross_validation"
        / f"fold_{fold}"
    )

    train_dataset = CariesDetectionDataset(
        FOLD_DIR / "images" / "train",
        FOLD_DIR / "labels" / "train"
    )

    val_dataset = CariesDetectionDataset(
        FOLD_DIR / "images" / "val",
        FOLD_DIR / "labels" / "val"
    )

    train_loader = DataLoader(
        train_dataset,
        batch_size=2,
        shuffle=True,
        num_workers=2,
        collate_fn=collate_fn
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=1,
        shuffle=False,
        num_workers=2,
        collate_fn=collate_fn
    )

    print("Train images:", len(train_dataset))
    print("Val images:", len(val_dataset))

    # =========================================================
    # FRESH MODEL
    # =========================================================

    model = fasterrcnn_resnet50_fpn(
        weights="DEFAULT"
    )

    in_features = (
        model.roi_heads
        .box_predictor
        .cls_score
        .in_features
    )

    model.roi_heads.box_predictor = FastRCNNPredictor(
        in_features,
        2
    )

    model = model.to(device)

    # =========================================================
    # OPTIMIZER
    # =========================================================

    params = [
        p for p in model.parameters()
        if p.requires_grad
    ]

    optimizer = torch.optim.AdamW(
        params,
        lr=1e-4,
        weight_decay=1e-4
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=3
    )

    # =========================================================
    # TRAINING
    # =========================================================

    max_epochs = 40
    early_stop_patience = 8

    best_val_loss = float("inf")
    best_state = None
    epochs_without_improvement = 0

    for epoch in range(max_epochs):

        # TRAIN
        model.train()

        train_loss_total = 0.0

        for images, targets in train_loader:

            images = [
                img.to(device)
                for img in images
            ]

            targets = [
                {
                    k: v.to(device)
                    for k, v in t.items()
                }
                for t in targets
            ]

            loss_dict = model(
                images,
                targets
            )

            losses = sum(
                loss
                for loss in loss_dict.values()
            )

            optimizer.zero_grad()
            losses.backward()
            optimizer.step()

            train_loss_total += losses.item()

        avg_train_loss = (
            train_loss_total /
            len(train_loader)
        )

        # VALIDATION LOSS
        model.train()

        val_loss_total = 0.0

        with torch.no_grad():

            for images, targets in val_loader:

                images = [
                    img.to(device)
                    for img in images
                ]

                targets = [
                    {
                        k: v.to(device)
                        for k, v in t.items()
                    }
                    for t in targets
                ]

                loss_dict = model(
                    images,
                    targets
                )

                losses = sum(
                    loss
                    for loss in loss_dict.values()
                )

                val_loss_total += losses.item()

        avg_val_loss = (
            val_loss_total /
            len(val_loader)
        )

        scheduler.step(
            avg_val_loss
        )

        print(
            f"Epoch {epoch+1:02d}/{max_epochs} | "
            f"Train Loss: {avg_train_loss:.4f} | "
            f"Val Loss: {avg_val_loss:.4f}"
        )

        if avg_val_loss < best_val_loss:

            best_val_loss = avg_val_loss

            best_state = deepcopy(
                model.state_dict()
            )

            epochs_without_improvement = 0

            print("✅ New best model")

        else:

            epochs_without_improvement += 1

            print(
                "No improvement:",
                epochs_without_improvement
            )

        if epochs_without_improvement >= early_stop_patience:

            print("🛑 Early stopping")
            break

    # =========================================================
    # SAVE BEST MODEL
    # =========================================================

    FRCNN_DIR = (
        PROJECT_DIR
        / "cross_validation"
        / "models"
        / "FasterRCNN"
        / f"fold_{fold}"
    )

    FRCNN_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    best_model_path = (
        FRCNN_DIR /
        "best.pt"
    )

    torch.save(
        best_state,
        best_model_path
    )

    # =========================================================
    # LOAD BEST
    # =========================================================

    model.load_state_dict(
        best_state
    )

    model.eval()

    # =========================================================
    # mAP
    # =========================================================

    map_metric = MeanAveragePrecision(
        box_format="xyxy",
        iou_type="bbox"
    )

    predictions_all = []
    targets_all = []

    with torch.no_grad():

        for images, targets in val_loader:

            images_gpu = [
                img.to(device)
                for img in images
            ]

            outputs = model(
                images_gpu
            )

            for output, target in zip(
                outputs,
                targets
            ):

                predictions_all.append({
                    "boxes":
                        output["boxes"]
                        .detach()
                        .cpu(),

                    "scores":
                        output["scores"]
                        .detach()
                        .cpu(),

                    "labels":
                        output["labels"]
                        .detach()
                        .cpu()
                })

                targets_all.append({
                    "boxes":
                        target["boxes"]
                        .detach()
                        .cpu(),

                    "labels":
                        target["labels"]
                        .detach()
                        .cpu()
                })

    map_metric.update(
        predictions_all,
        targets_all
    )

    map_results = map_metric.compute()

    # =========================================================
    # TP / FP / FN
    # =========================================================

    CONF_THRESHOLD = 0.25
    IOU_THRESHOLD = 0.50

    total_TP = 0
    total_FP = 0
    total_FN = 0

    with torch.no_grad():

        for images, targets in val_loader:

            images_gpu = [
                img.to(device)
                for img in images
            ]

            outputs = model(
                images_gpu
            )

            for output, target in zip(
                outputs,
                targets
            ):

                scores = (
                    output["scores"]
                    .detach()
                    .cpu()
                    .numpy()
                )

                boxes = (
                    output["boxes"]
                    .detach()
                    .cpu()
                    .numpy()
                )

                keep = (
                    scores >= CONF_THRESHOLD
                )

                pred_boxes = boxes[keep]

                gt_boxes = (
                    target["boxes"]
                    .detach()
                    .cpu()
                    .numpy()
                )

                matched_gt = set()

                TP = 0
                FP = 0

                for pred_box in pred_boxes:

                    best_iou = 0
                    best_gt_index = None

                    for gt_index, gt_box in enumerate(
                        gt_boxes
                    ):

                        if gt_index in matched_gt:
                            continue

                        iou = calculate_iou(
                            pred_box,
                            gt_box
                        )

                        if iou > best_iou:

                            best_iou = iou
                            best_gt_index = gt_index

                    if (
                        best_iou >= IOU_THRESHOLD
                        and best_gt_index is not None
                    ):

                        TP += 1
                        matched_gt.add(
                            best_gt_index
                        )

                    else:
                        FP += 1

                FN = (
                    len(gt_boxes)
                    -
                    len(matched_gt)
                )

                total_TP += TP
                total_FP += FP
                total_FN += FN

    precision = (
        total_TP /
        (total_TP + total_FP)
        if total_TP + total_FP > 0
        else 0
    )

    recall = (
        total_TP /
        (total_TP + total_FN)
        if total_TP + total_FN > 0
        else 0
    )

    # =========================================================
    # SUMMARY
    # =========================================================

    fold_summary = {
        "model": "Faster R-CNN",
        "fold": fold,
        "precision": precision,
        "recall": recall,
        "mAP50": float(
            map_results["map_50"]
        ),
        "mAP50_95": float(
            map_results["map"]
        )
    }

    print("\n" + "=" * 60)
    print(f"FASTER R-CNN — FOLD {fold}")
    print("=" * 60)

    for key, value in fold_summary.items():
        print(key, ":", value)

    # =========================================================
    # SAVE CSV AFTER EACH FOLD
    # =========================================================

    current = pd.read_csv(
        frcnn_results_path
    )

    current = current[
        current["fold"] != fold
    ]

    current = pd.concat(
        [
            current,
            pd.DataFrame([fold_summary])
        ],
        ignore_index=True
    )

    current = (
        current
        .sort_values("fold")
        .reset_index(drop=True)
    )

    current.to_csv(
        frcnn_results_path,
        index=False
    )

    print(
        f"✅ Fold {fold} saved"
    )

    # GPU cleanup
    del model
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

```python
for fold in [2, 3, 4, 5]:

    print("\n" + "=" * 80)
    print(f"RETINANET — STARTING FOLD {fold}")
    print("=" * 80)

    # =========================================================
    # DATASET
    # =========================================================

    FOLD_DIR = (
        PROJECT_DIR
        / "cross_validation"
        / f"fold_{fold}"
    )

    train_dataset = CariesDetectionDataset(
        FOLD_DIR / "images" / "train",
        FOLD_DIR / "labels" / "train"
    )

    val_dataset = CariesDetectionDataset(
        FOLD_DIR / "images" / "val",
        FOLD_DIR / "labels" / "val"
    )

    train_loader = DataLoader(
        train_dataset,
        batch_size=2,
        shuffle=True,
        num_workers=2,
        collate_fn=collate_fn
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=1,
        shuffle=False,
        num_workers=2,
        collate_fn=collate_fn
    )

    print("Train images:", len(train_dataset))
    print("Validation images:", len(val_dataset))

    # =========================================================
    # FRESH MODEL
    # =========================================================

    model = retinanet_resnet50_fpn(
        weights="DEFAULT"
    )

    num_classes = 2

    num_anchors = (
        model.head
        .classification_head
        .num_anchors
    )

    in_channels = (
        model.backbone.out_channels
    )

    model.head.classification_head = RetinaNetClassificationHead(
        in_channels,
        num_anchors,
        num_classes
    )

    model = model.to(device)

    # =========================================================
    # OPTIMIZER
    # =========================================================

    params = [
        p for p in model.parameters()
        if p.requires_grad
    ]

    optimizer = torch.optim.AdamW(
        params,
        lr=1e-4,
        weight_decay=1e-4
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=3
    )

    # =========================================================
    # TRAINING
    # =========================================================

    max_epochs = 40
    early_stop_patience = 8

    best_val_loss = float("inf")
    best_state = None
    epochs_without_improvement = 0

    for epoch in range(max_epochs):

        model.train()

        train_loss_total = 0.0

        for images, targets in train_loader:

            images = [
                img.to(device)
                for img in images
            ]

            targets = [
                {
                    k: v.to(device)
                    for k, v in t.items()
                }
                for t in targets
            ]

            loss_dict = model(
                images,
                targets
            )

            losses = sum(
                loss
                for loss in loss_dict.values()
            )

            optimizer.zero_grad()
            losses.backward()
            optimizer.step()

            train_loss_total += losses.item()

        avg_train_loss = (
            train_loss_total /
            len(train_loader)
        )

        # -------------------------
        # Validation loss
        # -------------------------

        model.train()

        val_loss_total = 0.0

        with torch.no_grad():

            for images, targets in val_loader:

                images = [
                    img.to(device)
                    for img in images
                ]

                targets = [
                    {
                        k: v.to(device)
                        for k, v in t.items()
                    }
                    for t in targets
                ]

                loss_dict = model(
                    images,
                    targets
                )

                losses = sum(
                    loss
                    for loss in loss_dict.values()
                )

                val_loss_total += losses.item()

        avg_val_loss = (
            val_loss_total /
            len(val_loader)
        )

        scheduler.step(
            avg_val_loss
        )

        print(
            f"Epoch {epoch+1:02d}/{max_epochs} | "
            f"Train Loss: {avg_train_loss:.4f} | "
            f"Val Loss: {avg_val_loss:.4f}"
        )

        if avg_val_loss < best_val_loss:

            best_val_loss = avg_val_loss

            best_state = deepcopy(
                model.state_dict()
            )

            epochs_without_improvement = 0

            print("✅ New best model")

        else:

            epochs_without_improvement += 1

            print(
                "No improvement:",
                epochs_without_improvement
            )

        if epochs_without_improvement >= early_stop_patience:

            print("🛑 Early stopping")
            break

    # =========================================================
    # SAVE BEST MODEL
    # =========================================================

    RETINA_DIR = (
        PROJECT_DIR
        / "cross_validation"
        / "models"
        / "RetinaNet"
        / f"fold_{fold}"
    )

    RETINA_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    best_path = (
        RETINA_DIR /
        "best.pt"
    )

    torch.save(
        best_state,
        best_path
    )

    model.load_state_dict(
        best_state
    )

    model.eval()

    # =========================================================
    # mAP EVALUATION
    # =========================================================

    map_metric = MeanAveragePrecision(
        box_format="xyxy",
        iou_type="bbox"
    )

    predictions_all = []
    targets_all = []

    with torch.no_grad():

        for images, targets in val_loader:

            images_gpu = [
                img.to(device)
                for img in images
            ]

            outputs = model(
                images_gpu
            )

            for output, target in zip(
                outputs,
                targets
            ):

                predictions_all.append({
                    "boxes":
                        output["boxes"]
                        .detach()
                        .cpu(),

                    "scores":
                        output["scores"]
                        .detach()
                        .cpu(),

                    "labels":
                        output["labels"]
                        .detach()
                        .cpu()
                })

                targets_all.append({
                    "boxes":
                        target["boxes"]
                        .detach()
                        .cpu(),

                    "labels":
                        target["labels"]
                        .detach()
                        .cpu()
                })

    map_metric.update(
        predictions_all,
        targets_all
    )

    map_results = map_metric.compute()

    # =========================================================
    # THRESHOLD SWEEP
    # =========================================================

    thresholds = [
        0.01,
        0.025,
        0.05,
        0.075,
        0.10,
        0.125,
        0.15
    ]

    threshold_rows = []

    for threshold in thresholds:

        total_TP = 0
        total_FP = 0
        total_FN = 0

        with torch.no_grad():

            for images, targets in val_loader:

                images_gpu = [
                    img.to(device)
                    for img in images
                ]

                outputs = model(
                    images_gpu
                )

                for output, target in zip(
                    outputs,
                    targets
                ):

                    scores = (
                        output["scores"]
                        .detach()
                        .cpu()
                        .numpy()
                    )

                    boxes = (
                        output["boxes"]
                        .detach()
                        .cpu()
                        .numpy()
                    )

                    keep = scores >= threshold
                    pred_boxes = boxes[keep]

                    gt_boxes = (
                        target["boxes"]
                        .detach()
                        .cpu()
                        .numpy()
                    )

                    matched_gt = set()

                    TP = 0
                    FP = 0

                    for pred_box in pred_boxes:

                        best_iou = 0
                        best_gt_index = None

                        for gt_index, gt_box in enumerate(
                            gt_boxes
                        ):

                            if gt_index in matched_gt:
                                continue

                            iou = calculate_iou(
                                pred_box,
                                gt_box
                            )

                            if iou > best_iou:
                                best_iou = iou
                                best_gt_index = gt_index

                        if (
                            best_iou >= 0.50
                            and best_gt_index is not None
                        ):

                            TP += 1
                            matched_gt.add(
                                best_gt_index
                            )

                        else:
                            FP += 1

                    FN = (
                        len(gt_boxes)
                        -
                        len(matched_gt)
                    )

                    total_TP += TP
                    total_FP += FP
                    total_FN += FN

        precision = (
            total_TP /
            (total_TP + total_FP)
            if total_TP + total_FP > 0
            else 0
        )

        recall = (
            total_TP /
            (total_TP + total_FN)
            if total_TP + total_FN > 0
            else 0
        )

        f1 = (
            2 * precision * recall /
            (precision + recall)
            if precision + recall > 0
            else 0
        )

        threshold_rows.append({
            "threshold": threshold,
            "precision": precision,
            "recall": recall,
            "F1": f1
        })

    threshold_df = pd.DataFrame(
        threshold_rows
    )

    best_threshold_row = threshold_df.loc[
        threshold_df["F1"].idxmax()
    ]

    # =========================================================
    # FINAL FOLD RESULT
    # =========================================================

    fold_summary = {
        "model": "RetinaNet",
        "fold": fold,
        "precision": float(
            best_threshold_row["precision"]
        ),
        "recall": float(
            best_threshold_row["recall"]
        ),
        "mAP50": float(
            map_results["map_50"]
        ),
        "mAP50_95": float(
            map_results["map"]
        )
    }

    print("\n" + "=" * 60)
    print(f"RETINANET — FOLD {fold}")
    print("=" * 60)

    print(
        "Best threshold:",
        best_threshold_row["threshold"]
    )

    for key, value in fold_summary.items():
        print(key, ":", value)

    # =========================================================
    # SAVE RESULT
    # =========================================================

    current = pd.read_csv(
        retina_results_path
    )

    current = current[
        current["fold"] != fold
    ]

    current = pd.concat(
        [
            current,
            pd.DataFrame([fold_summary])
        ],
        ignore_index=True
    )

    current = (
        current
        .sort_values("fold")
        .reset_index(drop=True)
    )

    current.to_csv(
        retina_results_path,
        index=False
    )

    print(
        f"✅ Fold {fold} saved"
    )

    del model
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

```python
from scipy.stats import wilcoxon
from itertools import combinations

models = {
    "YOLO11s": yolo,
    "RT-DETR": rt,
    "Faster R-CNN": frcnn,
    "RetinaNet": retina
}

pairwise_results = []

for metric in ["mAP50", "mAP50_95"]:

    for model_a, model_b in combinations(
        models.keys(),
        2
    ):

        values_a = models[model_a][metric].values
        values_b = models[model_b][metric].values

        statistic, p_value = wilcoxon(
            values_a,
            values_b,
            alternative="two-sided"
        )

        pairwise_results.append({
            "Metric": metric,
            "Model A": model_a,
            "Model B": model_b,
            "Statistic": statistic,
            "p_raw": p_value
        })

pairwise_df = pd.DataFrame(
    pairwise_results
)

print(pairwise_df)
```

### Step 193: Apply Holm multiplicity correction

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
from statsmodels.stats.multitest import multipletests

pairwise_df["p_holm"] = None

for metric in [
    "mAP50",
    "mAP50_95"
]:

    mask = (
        pairwise_df["Metric"]
        == metric
    )

    p_values = (
        pairwise_df
        .loc[mask, "p_raw"]
        .values
    )

    corrected = multipletests(
        p_values,
        alpha=0.05,
        method="holm"
    )

    pairwise_df.loc[
        mask,
        "p_holm"
    ] = corrected[1]

pairwise_df["Significant"] = (
    pairwise_df["p_holm"]
    .astype(float)
    < 0.05
)

print(
    pairwise_df.to_string(
        index=False
    )
)
```

```python
CARIES_LABEL = "龋病"
CLASS_ID = 0

def convert_final_case(row):

    json_file = Path(
        row["json_path"]
    )

    with open(
        json_file,
        "r",
        encoding="utf-8"
    ) as f:

        data = json.load(f)

    image_path = find_image(
        json_file,
        data
    )

    if image_path is None:
        raise FileNotFoundError(
            f"Image not found: {json_file}"
        )

    W = data.get("imageWidth")
    H = data.get("imageHeight")

    if not W or not H:

        with Image.open(image_path) as img:
            W, H = img.size

    shutil.copy2(
        image_path,
        FINAL_DIR
        / "images"
        / "train"
        / image_path.name
    )

    lines = []

    for shape in data.get("shapes", []):

        if str(
            shape.get("label", "")
        ).strip() != CARIES_LABEL:
            continue

        points = shape["points"]

        x1 = min(
            points[0][0],
            points[1][0]
        )

        y1 = min(
            points[0][1],
            points[1][1]
        )

        x2 = max(
            points[0][0],
            points[1][0]
        )

        y2 = max(
            points[0][1],
            points[1][1]
        )

        bw = x2 - x1
        bh = y2 - y1

        xc = (x1 + x2) / 2
        yc = (y1 + y2) / 2

        lines.append(
            f"0 "
            f"{xc/W:.6f} "
            f"{yc/H:.6f} "
            f"{bw/W:.6f} "
            f"{bh/H:.6f}"
        )

    label_path = (
        FINAL_DIR
        / "labels"
        / "train"
        / f"{image_path.stem}.txt"
    )

    label_path.write_text(
        "\n".join(lines)
    )

    return len(lines)
```

### Step 201: Define reusable analysis functions

```python
def find_image(json_file, data):

    json_file = Path(json_file)

    image_name = Path(
        str(data["imagePath"])
    ).name

    current = json_file.parent

    while current.name not in ["Train", "Test"]:

        if current == current.parent:
            return None

        current = current.parent

    direct_path = (
        current
        / "images"
        / image_name
    )

    if direct_path.exists():
        return direct_path

    image_stem = Path(image_name).stem

    for ext in [".png", ".jpg", ".jpeg", ".bmp"]:

        candidate = (
            current
            / "images"
            / f"{image_stem}{ext}"
        )

        if candidate.exists():
            return candidate

    return None
```

### Step 221: Define reusable analysis functions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
def calculate_iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])

    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = (
        max(0, x2 - x1)
        *
        max(0, y2 - y1)
    )

    area1 = (
        (box1[2] - box1[0])
        *
        (box1[3] - box1[1])
    )

    area2 = (
        (box2[2] - box2[0])
        *
        (box2[3] - box2[1])
    )

    union = area1 + area2 - intersection

    if union <= 0:
        return 0.0

    return intersection / union
```

### Step 222: Define reusable analysis functions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
from PIL import Image

def load_yolo_ground_truth(label_path, W, H):

    boxes = []

    if not label_path.exists():
        return boxes

    text = label_path.read_text().strip()

    if not text:
        return boxes

    for line in text.splitlines():

        cls, xc, yc, bw, bh = map(
            float,
            line.split()
        )

        xc *= W
        yc *= H
        bw *= W
        bh *= H

        x1 = xc - bw / 2
        y1 = yc - bh / 2
        x2 = xc + bw / 2
        y2 = yc + bh / 2

        boxes.append(
            [x1, y1, x2, y2]
        )

    return boxes
```

### Step 223: Generate model predictions

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
import pandas as pd

CONF_THRESHOLD = 0.25
IOU_THRESHOLD = 0.50

test_image_files = sorted([
    p for p in test_images_dir.iterdir()
    if p.suffix.lower() in [
        ".png", ".jpg", ".jpeg", ".bmp"
    ]
])

error_rows = []

for image_path in test_image_files:

    with Image.open(image_path) as img:
        W, H = img.size

    # Ground truth
    label_path = (
        test_labels_dir /
        f"{image_path.stem}.txt"
    )

    gt_boxes = load_yolo_ground_truth(
        label_path,
        W,
        H
    )

    # Predictions
    prediction = final_detector.predict(
        source=str(image_path),
        imgsz=1024,
        conf=CONF_THRESHOLD,
        iou=0.50,
        device=0,
        verbose=False
    )[0]

    pred_boxes = []

    if prediction.boxes is not None:

        for box in prediction.boxes:

            coords = (
                box.xyxy[0]
                .detach()
                .cpu()
                .numpy()
                .tolist()
            )

            score = float(
                box.conf[0]
                .detach()
                .cpu()
            )

            pred_boxes.append(
                {
                    "box": coords,
                    "score": score
                }
            )

    # Matching
    matched_gt = set()

    TP = 0
    FP = 0

    for pred in pred_boxes:

        best_iou = 0
        best_gt_index = None

        for gt_index, gt_box in enumerate(
            gt_boxes
        ):

            if gt_index in matched_gt:
                continue

            iou = calculate_iou(
                pred["box"],
                gt_box
            )

            if iou > best_iou:
                best_iou = iou
                best_gt_index = gt_index

        if (
            best_iou >= IOU_THRESHOLD
            and best_gt_index is not None
        ):

            TP += 1
            matched_gt.add(
                best_gt_index
            )

        else:
            FP += 1

    FN = (
        len(gt_boxes)
        -
        len(matched_gt)
    )

    error_rows.append({
        "image": image_path.name,
        "GT": len(gt_boxes),
        "Predictions": len(pred_boxes),
        "TP": TP,
        "FP": FP,
        "FN": FN
    })


error_df = pd.DataFrame(
    error_rows
)

print(
    error_df.to_string(
        index=False
    )
)
```

### Step 230: Generate model predictions

Creates a visual quality-control or publication figure from the variables prepared in the preceding cells.


```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
from pathlib import Path

# ==========================================================
# الحالات المختارة من الـ Final Test
# ==========================================================

selected_cases = [
    ("cate1_011.png", "Good detection"),
    ("cate2_004.png", "Mixed detection"),
    ("cate1_007.png", "False-negative dominant"),
    ("cate1_000.png", "False-positive case")
]

CONF_THRESHOLD = 0.25
IOU_THRESHOLD = 0.50


fig, axes = plt.subplots(
    2,
    2,
    figsize=(20, 11)
)


for ax, (image_name, case_title) in zip(
    axes.flatten(),
    selected_cases
):

    # ------------------------------------------------------
    # Load image
    # ------------------------------------------------------

    image_path = (
        test_images_dir /
        image_name
    )

    label_path = (
        test_labels_dir /
        f"{Path(image_name).stem}.txt"
    )

    img = Image.open(image_path)

    W, H = img.size

    ax.imshow(
        img,
        cmap="gray"
    )

    # ------------------------------------------------------
    # Ground Truth
    # ------------------------------------------------------

    gt_boxes = load_yolo_ground_truth(
        label_path,
        W,
        H
    )

    # ------------------------------------------------------
    # Predictions
    # ------------------------------------------------------

    prediction = final_detector.predict(
        source=str(image_path),
        imgsz=1024,
        conf=CONF_THRESHOLD,
        iou=0.50,
        device=0,
        verbose=False
    )[0]

    pred_boxes = []

    if prediction.boxes is not None:

        for box in prediction.boxes:

            coords = (
                box.xyxy[0]
                .detach()
                .cpu()
                .numpy()
                .tolist()
            )

            score = float(
                box.conf[0]
                .detach()
                .cpu()
            )

            pred_boxes.append({
                "box": coords,
                "score": score
            })

    # ------------------------------------------------------
    # Match predictions to GT
    # ------------------------------------------------------

    matched_gt = set()
    evaluated_predictions = []

    for pred in pred_boxes:

        best_iou = 0
        best_gt_index = None

        for gt_index, gt_box in enumerate(
            gt_boxes
        ):

            if gt_index in matched_gt:
                continue

            iou = calculate_iou(
                pred["box"],
                gt_box
            )

            if iou > best_iou:

                best_iou = iou
                best_gt_index = gt_index

        # True Positive
        if (
            best_iou >= IOU_THRESHOLD
            and best_gt_index is not None
        ):

            matched_gt.add(
                best_gt_index
            )

            evaluated_predictions.append({
                "box": pred["box"],
                "score": pred["score"],
                "type": "TP"
            })

        # False Positive
        else:

            evaluated_predictions.append({
                "box": pred["box"],
                "score": pred["score"],
                "type": "FP"
            })

    # ------------------------------------------------------
    # 1. All Ground Truth = RED
    # ------------------------------------------------------

    for gt_box in gt_boxes:

        x1, y1, x2, y2 = gt_box

        rect = patches.Rectangle(
            (x1, y1),
            x2 - x1,
            y2 - y1,
            linewidth=1.5,
            edgecolor="red",
            facecolor="none"
        )

        ax.add_patch(rect)

    # ------------------------------------------------------
    # 2. TP predictions = GREEN
    #    FP predictions = ORANGE
    # ------------------------------------------------------

    for pred in evaluated_predictions:

        x1, y1, x2, y2 = pred["box"]

        if pred["type"] == "TP":

            edge_color = "lime"
            label_text = f"TP {pred['score']:.2f}"

        else:

            edge_color = "orange"
            label_text = f"FP {pred['score']:.2f}"

        rect = patches.Rectangle(
            (x1, y1),
            x2 - x1,
            y2 - y1,
            linewidth=2.3,
            edgecolor=edge_color,
            facecolor="none"
        )

        ax.add_patch(rect)

        ax.text(
            x1,
            max(0, y1 - 8),
            label_text,
            fontsize=9,
            bbox=dict(
                facecolor="white",
                alpha=0.8
            )
        )

    # ------------------------------------------------------
    # 3. Missed Ground Truth = BLUE = FN
    # ------------------------------------------------------

    for gt_index, gt_box in enumerate(
        gt_boxes
    ):

        if gt_index not in matched_gt:

            x1, y1, x2, y2 = gt_box

            rect = patches.Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                linewidth=3,
                edgecolor="blue",
                facecolor="none",
                linestyle="--"
            )

            ax.add_patch(rect)

            ax.text(
                x1,
                min(H - 10, y2 + 20),
                "FN",
                fontsize=9,
                bbox=dict(
                    facecolor="white",
                    alpha=0.8
                )
            )

    # ------------------------------------------------------
    # Statistics for image
    # ------------------------------------------------------

    row = error_df[
        error_df["image"] == image_name
    ].iloc[0]

    ax.set_title(
        f"{case_title}\n"
        f"{image_name} | "
        f"TP={int(row['TP'])}, "
        f"FP={int(row['FP'])}, "
        f"FN={int(row['FN'])}",
        fontsize=12
    )

    ax.axis("off")


plt.tight_layout()
plt.show()
```

### Step 255: Define reusable analysis functions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
def load_yolo_ground_truth(
    label_path,
    image_width,
    image_height
):

    boxes = []

    if not label_path.exists():
        return boxes

    text = (
        label_path
        .read_text()
        .strip()
    )

    if not text:
        return boxes


    for line in text.splitlines():

        values = line.split()

        cls = int(
            float(values[0])
        )

        xc = float(values[1])
        yc = float(values[2])
        bw = float(values[3])
        bh = float(values[4])


        # Convert normalized YOLO coordinates to pixels

        xc *= image_width
        yc *= image_height

        bw *= image_width
        bh *= image_height


        x1 = xc - bw / 2
        y1 = yc - bh / 2

        x2 = xc + bw / 2
        y2 = yc + bh / 2


        boxes.append(
            [
                x1,
                y1,
                x2,
                y2
            ]
        )

    return boxes
```

### Step 256: Define reusable analysis functions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
def calculate_iou(
    box1,
    box2
):

    x1 = max(
        box1[0],
        box2[0]
    )

    y1 = max(
        box1[1],
        box2[1]
    )

    x2 = min(
        box1[2],
        box2[2]
    )

    y2 = min(
        box1[3],
        box2[3]
    )


    intersection = (
        max(0, x2 - x1)
        *
        max(0, y2 - y1)
    )


    area1 = (
        max(0, box1[2] - box1[0])
        *
        max(0, box1[3] - box1[1])
    )


    area2 = (
        max(0, box2[2] - box2[0])
        *
        max(0, box2[3] - box2[1])
    )


    union = (
        area1 +
        area2 -
        intersection
    )


    if union <= 0:
        return 0.0


    return (
        intersection /
        union
    )
```

### Step 259: Estimate bootstrap confidence intervals

Resamples complete images, rather than individual lesions, to preserve within-image dependence while estimating uncertainty.


```python
# ============================================================
# IMAGE-LEVEL BOOTSTRAP
# Fixed-threshold Precision / Recall / F1
# ============================================================

B = 2000
SEED = 42

rng = np.random.default_rng(SEED)

n_images = len(df)

bootstrap_results = []

for b in range(B):

    # Sample 30 radiographs WITH replacement
    sampled_indices = rng.integers(
        0,
        n_images,
        size=n_images
    )

    sample = df.iloc[
        sampled_indices
    ]

    TP = sample["TP"].sum()
    FP = sample["FP"].sum()
    FN = sample["FN"].sum()

    precision = (
        TP / (TP + FP)
        if (TP + FP) > 0
        else np.nan
    )

    recall = (
        TP / (TP + FN)
        if (TP + FN) > 0
        else np.nan
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if (
            precision + recall > 0
        )
        else np.nan
    )

    bootstrap_results.append(
        [
            precision,
            recall,
            f1
        ]
    )


bootstrap_results = np.array(
    bootstrap_results
)

metrics = [
    "Precision",
    "Recall",
    "F1"
]

point_estimates = [
    total_tp / (total_tp + total_fp) if (total_tp + total_fp) else 0.0,
    total_tp / (total_tp + total_fn) if (total_tp + total_fn) else 0.0,
    (2 * total_tp) / (2 * total_tp + total_fp + total_fn)
    if (2 * total_tp + total_fp + total_fn) else 0.0,
]

summary_rows = []

for i, metric in enumerate(metrics):

    values = bootstrap_results[:, i]

    lower = np.nanpercentile(
        values,
        2.5
    )

    upper = np.nanpercentile(
        values,
        97.5
    )

    summary_rows.append({
        "Metric": metric,
        "Point_Estimate": point_estimates[i],
        "CI_2.5": lower,
        "CI_97.5": upper
    })


bootstrap_summary = pd.DataFrame(
    summary_rows
)

print(
    bootstrap_summary.round(4)
)


SAVE_PATH = (
    PROJECT_DIR /
    "final_model" /
    "YOLO11s_FixedThreshold_Bootstrap_CI.csv"
)

bootstrap_summary.to_csv(
    SAVE_PATH,
    index=False
)

print("\n✅ Saved:")
print(SAVE_PATH)
```

```python
import torch
from pathlib import Path
from PIL import Image
from ultralytics import YOLO
from torchmetrics.detection.mean_ap import MeanAveragePrecision

# ============================================================
# PATHS
# ============================================================

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Pediatric_Caries_AI"
)

MODEL_PATH = (
    PROJECT_DIR /
    "final_model" /
    "YOLO11s_Final" /
    "final_60epochs.pt"
)

DATA_DIR = (
    PROJECT_DIR /
    "processed_data" /
    "yolo_caries_baseline"
)

TEST_IMAGES = (
    DATA_DIR /
    "images" /
    "test"
)

TEST_LABELS = (
    DATA_DIR /
    "labels" /
    "test"
)

print("Model exists:", MODEL_PATH.exists())
print("Images exist:", TEST_IMAGES.exists())
print("Labels exist:", TEST_LABELS.exists())


# ============================================================
# LOAD MODEL
# ============================================================

model = YOLO(
    str(MODEL_PATH)
)


# ============================================================
# GROUND-TRUTH LOADER
# ============================================================

def load_target(
    label_path,
    width,
    height
):

    boxes = []

    if label_path.exists():

        text = label_path.read_text().strip()

        if text:

            for line in text.splitlines():

                values = line.split()

                xc = float(values[1]) * width
                yc = float(values[2]) * height
                bw = float(values[3]) * width
                bh = float(values[4]) * height

                x1 = xc - bw / 2
                y1 = yc - bh / 2
                x2 = xc + bw / 2
                y2 = yc + bh / 2

                boxes.append(
                    [x1, y1, x2, y2]
                )

    if len(boxes) > 0:

        boxes = torch.tensor(
            boxes,
            dtype=torch.float32
        )

    else:

        boxes = torch.empty(
            (0, 4),
            dtype=torch.float32
        )

    labels = torch.zeros(
        len(boxes),
        dtype=torch.int64
    )

    return {
        "boxes": boxes,
        "labels": labels
    }


# ============================================================
# GET TEST IMAGES
# ============================================================

image_paths = sorted(
    list(TEST_IMAGES.glob("*.png")) +
    list(TEST_IMAGES.glob("*.jpg")) +
    list(TEST_IMAGES.glob("*.jpeg"))
)

print("\nTest images found:", len(image_paths))


# ============================================================
# CACHE PREDICTIONS AND TARGETS
# ============================================================

prediction_cache = []
target_cache = []


for i, image_path in enumerate(image_paths, start=1):

    img = Image.open(
        image_path
    )

    W, H = img.size

    target = load_target(
        TEST_LABELS /
        f"{image_path.stem}.txt",
        W,
        H
    )

    result = model.predict(
        source=str(image_path),

        imgsz=1024,

        # Important for AP calculation
        conf=0.001,

        # NMS IoU setting
        iou=0.70,

        max_det=300,

        device=0,

        verbose=False
    )[0]


    if (
        result.boxes is not None
        and len(result.boxes) > 0
    ):

        pred = {

            "boxes":
                result.boxes.xyxy
                .detach()
                .cpu()
                .float(),

            "scores":
                result.boxes.conf
                .detach()
                .cpu()
                .float(),

            "labels":
                result.boxes.cls
                .detach()
                .cpu()
                .long()
        }

    else:

        pred = {

            "boxes":
                torch.empty(
                    (0, 4),
                    dtype=torch.float32
                ),

            "scores":
                torch.empty(
                    (0,),
                    dtype=torch.float32
                ),

            "labels":
                torch.empty(
                    (0,),
                    dtype=torch.int64
                )
        }


    prediction_cache.append(
        pred
    )

    target_cache.append(
        target
    )

    print(
        f"{i:02d}/{len(image_paths)}",
        image_path.name,
        "| GT:",
        len(target["boxes"]),
        "| Pred:",
        len(pred["boxes"])
    )


print("\n✅ prediction_cache created:", len(prediction_cache))
print("✅ target_cache created:", len(target_cache))


# ============================================================
# QUALITY CHECK
# ============================================================

metric = MeanAveragePrecision(
    box_format="xyxy",
    iou_type="bbox",
    max_detection_thresholds=[
        1,
        10,
        300
    ]
)

metric.update(
    prediction_cache,
    target_cache
)

base_result = metric.compute()

custom_map50 = float(
    base_result["map_50"]
)

custom_map5095 = float(
    base_result["map"]
)


OFFICIAL_MAP50 = 0.459214
OFFICIAL_MAP5095 = 0.247294


print("\n===================================")
print("QUALITY CHECK")
print("===================================")

print(
    "Custom mAP@0.50       :",
    round(custom_map50, 6)
)

print(
    "Official mAP@0.50     :",
    OFFICIAL_MAP50
)

print(
    "Difference            :",
    round(
        custom_map50 -
        OFFICIAL_MAP50,
        6
    )
)

print()

print(
    "Custom mAP@0.50:0.95  :",
    round(custom_map5095, 6)
)

print(
    "Official mAP@0.50:0.95:",
    OFFICIAL_MAP5095
)

print(
    "Difference            :",
    round(
        custom_map5095 -
        OFFICIAL_MAP5095,
        6
    )
)
```

```python
import copy
from pathlib import Path
from ultralytics import YOLO
from ultralytics.models.yolo.detect.val import DetectionValidator

# ============================================================
# PATHS
# ============================================================

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Pediatric_Caries_AI"
)

MODEL_PATH = (
    PROJECT_DIR /
    "final_model" /
    "YOLO11s_Final" /
    "final_60epochs.pt"
)

DATA_DIR = (
    PROJECT_DIR /
    "processed_data" /
    "yolo_caries_baseline"
)

# ============================================================
# FIND DATA YAML
# ============================================================

yaml_candidates = list(
    DATA_DIR.rglob("*.yaml")
) + list(
    DATA_DIR.rglob("*.yml")
)

print("YAML candidates:")

for y in yaml_candidates:
    print(" -", y)

if len(yaml_candidates) == 0:
    raise FileNotFoundError(
        "No YAML file found inside the YOLO dataset directory."
    )

# Prefer YAML containing a test split
DATA_YAML = None

for y in yaml_candidates:

    try:
        txt = y.read_text(
            encoding="utf-8"
        )

        if "test:" in txt:
            DATA_YAML = y
            break

    except:
        pass

if DATA_YAML is None:
    DATA_YAML = yaml_candidates[0]

print("\n✅ Using YAML:")
print(DATA_YAML)


# ============================================================
# STORAGE FOR EXACT VALIDATOR STATS
# ============================================================

captured = {}

original_get_stats = DetectionValidator.get_stats


def patched_get_stats(self):

    # --------------------------------------------------------
    # Capture stats BEFORE Ultralytics processes / clears them
    # --------------------------------------------------------

    captured["stats"] = {
        key: [
            copy.deepcopy(item)
            for item in value
        ]
        for key, value
        in self.metrics.stats.items()
    }

    captured["names"] = copy.deepcopy(
        self.metrics.names
    )

    # Continue normal official validation
    return original_get_stats(self)


# ============================================================
# TEMPORARILY PATCH VALIDATOR
# ============================================================

DetectionValidator.get_stats = patched_get_stats

try:

    exact_model = YOLO(
        str(MODEL_PATH)
    )

    official_result = exact_model.val(

        data=str(DATA_YAML),

        split="test",

        imgsz=1024,

        # Ultralytics AP evaluation normally uses
        # a very low confidence threshold
        conf=0.001,

        iou=0.70,

        max_det=300,

        device=0,

        plots=False,

        save=False,

        verbose=False
    )

finally:

    # ALWAYS restore original function
    DetectionValidator.get_stats = (
        original_get_stats
    )


# ============================================================
# OFFICIAL RESULTS FROM THIS RUN
# ============================================================

print()
print("==========================================")
print("OFFICIAL VALIDATOR RESULT")
print("==========================================")

print(
    "Precision        :",
    round(
        float(official_result.box.mp),
        6
    )
)

print(
    "Recall           :",
    round(
        float(official_result.box.mr),
        6
    )
)

print(
    "mAP@0.50         :",
    round(
        float(official_result.box.map50),
        6
    )
)

print(
    "mAP@0.50:0.95    :",
    round(
        float(official_result.box.map),
        6
    )
)


# ============================================================
# CHECK CAPTURED PER-IMAGE DATA
# ============================================================

print()
print("==========================================")
print("CAPTURED STATS")
print("==========================================")

if "stats" not in captured:
    raise RuntimeError(
        "Validator statistics were not captured."
    )

for key, value in captured["stats"].items():
    print(
        key,
        ":",
        len(value)
    )
```

### Step 263: Estimate bootstrap confidence intervals

Resamples complete images, rather than individual lesions, to preserve within-image dependence while estimating uncertainty.


```python
import copy
import time
import numpy as np
import pandas as pd

from pathlib import Path
from ultralytics.utils.metrics import DetMetrics


# ============================================================
# SETTINGS
# ============================================================

B = 2000
SEED = 42

rng = np.random.default_rng(SEED)

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Pediatric_Caries_AI"
)

OUTPUT_DIR = (
    PROJECT_DIR /
    "final_model"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# RECONSTRUCT THE 30 OFFICIAL PER-IMAGE STAT DICTIONARIES
# ============================================================

captured_stats = captured["stats"]

stat_keys = list(
    captured_stats.keys()
)

n_images = len(
    captured_stats["tp"]
)

assert n_images == 30


official_per_image_stats = []

for i in range(n_images):

    stat = {}

    for key in stat_keys:

        stat[key] = copy.deepcopy(
            captured_stats[key][i]
        )

    official_per_image_stats.append(
        stat
    )


print(
    "✅ Official per-image stats reconstructed:",
    len(official_per_image_stats)
)


# ============================================================
# FUNCTION:
# EVALUATE ANY RESAMPLED SET USING ULTRALYTICS DetMetrics
# ============================================================

def evaluate_bootstrap_indices(indices):

    metrics = DetMetrics(
        names={
            0: "caries"
        }
    )

    # --------------------------------------------------------
    # Populate metrics.stats directly from the
    # official captured validator statistics
    # --------------------------------------------------------

    new_stats = {}

    for key in stat_keys:

        new_stats[key] = [
            copy.deepcopy(
                official_per_image_stats[idx][key]
            )
            for idx in indices
        ]

    metrics.stats = new_stats


    # --------------------------------------------------------
    # Official Ultralytics AP processing
    # --------------------------------------------------------

    metrics.process(
        save_dir=Path("/content"),
        plot=False,
        on_plot=None
    )


    mean_results = metrics.mean_results()

    precision = float(
        mean_results[0]
    )

    recall = float(
        mean_results[1]
    )

    map50 = float(
        mean_results[2]
    )

    map5095 = float(
        mean_results[3]
    )


    if (
        precision + recall
    ) > 0:

        f1 = (
            2 *
            precision *
            recall /
            (
                precision +
                recall
            )
        )

    else:

        f1 = np.nan


    return (
        precision,
        recall,
        f1,
        map50,
        map5095
    )


# ============================================================
# QUALITY CHECK BEFORE BOOTSTRAP
# Evaluate original 30 images ONCE
# ============================================================

baseline = evaluate_bootstrap_indices(
    np.arange(n_images)
)

baseline_precision = baseline[0]
baseline_recall = baseline[1]
baseline_f1 = baseline[2]
baseline_map50 = baseline[3]
baseline_map5095 = baseline[4]


# Use exact official validator values if object
# from previous cell still exists
try:

    official_precision = float(
        official_result.box.mp
    )

    official_recall = float(
        official_result.box.mr
    )

    official_map50 = float(
        official_result.box.map50
    )

    official_map5095 = float(
        official_result.box.map
    )

except:

    official_precision = 0.427713
    official_recall = 0.601852
    official_map50 = 0.459214
    official_map5095 = 0.247294


official_f1 = (
    2 *
    official_precision *
    official_recall /
    (
        official_precision +
        official_recall
    )
)


print()
print("==============================================")
print("FINAL BASELINE CHECK")
print("==============================================")

print(
    "Precision      :",
    round(baseline_precision, 6),
    "| Official:",
    round(official_precision, 6)
)

print(
    "Recall         :",
    round(baseline_recall, 6),
    "| Official:",
    round(official_recall, 6)
)

print(
    "F1             :",
    round(baseline_f1, 6),
    "| Official:",
    round(official_f1, 6)
)

print(
    "mAP@0.50       :",
    round(baseline_map50, 6),
    "| Official:",
    round(official_map50, 6)
)

print(
    "mAP@0.50:0.95  :",
    round(baseline_map5095, 6),
    "| Official:",
    round(official_map5095, 6)
)


# ============================================================
# SAFETY CHECK
# ============================================================

map50_difference = abs(
    baseline_map50 -
    official_map50
)

map5095_difference = abs(
    baseline_map5095 -
    official_map5095
)


if (
    map50_difference > 0.0001
    or
    map5095_difference > 0.0001
):

    raise RuntimeError(
        "Baseline reconstruction does not match "
        "the official Ultralytics result closely enough. "
        "Bootstrap stopped."
    )


print()
print("✅ Baseline matched. Starting 2,000 bootstrap replicates...")


# ============================================================
# 2,000 IMAGE-LEVEL BOOTSTRAP REPLICATES
# ============================================================

bootstrap_rows = []

start_time = time.time()


for b in range(B):

    # --------------------------------------------------------
    # Resample 30 RADIOGRAPHS with replacement
    # --------------------------------------------------------

    sampled_indices = rng.integers(
        low=0,
        high=n_images,
        size=n_images
    )


    (
        precision,
        recall,
        f1,
        map50,
        map5095
    ) = evaluate_bootstrap_indices(
        sampled_indices
    )


    bootstrap_rows.append({
        "replicate": b + 1,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "mAP50": map50,
        "mAP50_95": map5095
    })


    if (
        (b + 1) % 200 == 0
    ):

        elapsed = (
            time.time() -
            start_time
        )

        print(
            f"✅ {b + 1}/{B} completed "
            f"| elapsed: {elapsed:.1f} s"
        )


# ============================================================
# CREATE RAW BOOTSTRAP DATAFRAME
# ============================================================

bootstrap_df = pd.DataFrame(
    bootstrap_rows
)


# ============================================================
# SAVE ALL 2,000 REPLICATES
# ============================================================

RAW_PATH = (
    OUTPUT_DIR /
    "YOLO11s_PrincipalMetrics_Bootstrap_2000.csv"
)

bootstrap_df.to_csv(
    RAW_PATH,
    index=False
)


# ============================================================
# CALCULATE 95% PERCENTILE CONFIDENCE INTERVALS
# ============================================================

point_estimates = {

    "Precision":
        official_precision,

    "Recall":
        official_recall,

    "F1":
        official_f1,

    "mAP50":
        official_map50,

    "mAP50_95":
        official_map5095
}


summary_rows = []


for metric_name in [

    "Precision",
    "Recall",
    "F1",
    "mAP50",
    "mAP50_95"

]:

    values = (
        bootstrap_df[
            metric_name
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
        .values
    )


    ci_lower = np.percentile(
        values,
        2.5
    )

    ci_upper = np.percentile(
        values,
        97.5
    )


    bootstrap_mean = np.mean(
        values
    )


    bootstrap_median = np.median(
        values
    )


    summary_rows.append({

        "Metric":
            metric_name,

        "Point_Estimate":
            point_estimates[
                metric_name
            ],

        "Bootstrap_Mean":
            bootstrap_mean,

        "Bootstrap_Median":
            bootstrap_median,

        "CI_2.5":
            ci_lower,

        "CI_97.5":
            ci_upper,

        "Replicates":
            len(values)
    })


bootstrap_summary = pd.DataFrame(
    summary_rows
)


# ============================================================
# SAVE SUMMARY
# ============================================================

SUMMARY_PATH = (
    OUTPUT_DIR /
    "YOLO11s_PrincipalMetrics_Bootstrap_CI.csv"
)

bootstrap_summary.to_csv(
    SUMMARY_PATH,
    index=False
)


# ============================================================
# DISPLAY FINAL RESULTS
# ============================================================

print()
print()
print("========================================================")
print("FINAL 95% IMAGE-LEVEL BOOTSTRAP CONFIDENCE INTERVALS")
print("========================================================")

for _, row in bootstrap_summary.iterrows():

    print(
        f"{row['Metric']:12s} "
        f"= {row['Point_Estimate']:.4f} "
        f"(95% CI {row['CI_2.5']:.4f}–"
        f"{row['CI_97.5']:.4f})"
    )


print()
print("✅ Raw bootstrap replicates saved:")
print(RAW_PATH)

print()
print("✅ Bootstrap summary saved:")
print(SUMMARY_PATH)
```

### Step 289: Estimate bootstrap confidence intervals

Resamples complete images, rather than individual lesions, to preserve within-image dependence while estimating uncertainty.


```python
import copy
import numpy as np
import pandas as pd

from pathlib import Path
from ultralytics.utils.metrics import DetMetrics

B = 2000
SEED = 42

rng = np.random.default_rng(SEED)

stats = captured_external["stats"]
keys = list(stats.keys())

n_images = len(stats["tp"])

per_image = []

for i in range(n_images):

    d = {}

    for key in keys:
        d[key] = copy.deepcopy(
            stats[key][i]
        )

    per_image.append(d)


def evaluate_indices(indices):

    m = DetMetrics(
        names={0: "caries"}
    )

    new_stats = {}

    for key in keys:

        new_stats[key] = [
            copy.deepcopy(
                per_image[i][key]
            )
            for i in indices
        ]

    m.stats = new_stats

    m.process(
        save_dir=Path("/content"),
        plot=False,
        on_plot=None
    )

    p, r, map50, map5095 = m.mean_results()

    p = float(p)
    r = float(r)
    map50 = float(map50)
    map5095 = float(map5095)

    f1 = (
        2 * p * r / (p + r)
        if (p + r) > 0
        else 0.0
    )

    return p, r, f1, map50, map5095


# Baseline reconstruction
baseline = evaluate_indices(
    np.arange(n_images)
)

print("Baseline:", baseline)


rows = []

for b in range(B):

    idx = rng.integers(
        0,
        n_images,
        size=n_images
    )

    p, r, f1, m50, m95 = evaluate_indices(idx)

    rows.append({
        "replicate": b + 1,
        "Precision": p,
        "Recall": r,
        "F1": f1,
        "mAP50": m50,
        "mAP50_95": m95
    })


df = pd.DataFrame(rows)

output_raw = (
    EXTERNAL_DIR /
    "external_bootstrap_2000.csv"
)

df.to_csv(
    output_raw,
    index=False
)


for metric in [
    "Precision",
    "Recall",
    "F1",
    "mAP50",
    "mAP50_95"
]:

    values = df[metric].values

    low = np.percentile(values, 2.5)
    high = np.percentile(values, 97.5)

    print(
        metric,
        ":",
        round(low, 4),
        "–",
        round(high, 4)
    )
```

### Step 290: Generate model predictions

Builds the current result table from computed variables and writes it to the project output directory.

> Interpretation note: image-level positivity can be nonzero even when lesion-level TP is zero because it ignores localization overlap.


```python
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from torchvision.ops import box_iou


CONF_THR = 0.25
IOU_THR = 0.50

images_dir = EXTERNAL_DIR / "images" / "test"
labels_dir = EXTERNAL_DIR / "labels" / "test"

image_files = sorted([
    p for p in images_dir.iterdir()
    if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
])


# ------------------------------------------------------------
# Get predictions once at low confidence
# ------------------------------------------------------------

preds_ext = model.predict(
    source=[str(p) for p in image_files],
    imgsz=1024,
    conf=0.001,
    iou=0.70,
    max_det=300,
    verbose=False
)


total_TP = 0
total_FP = 0
total_FN = 0

img_TP = 0
img_FP = 0
img_TN = 0
img_FN = 0

rows = []


for r in preds_ext:

    stem = Path(r.path).stem
    img_h, img_w = r.orig_shape

    # ========================================================
    # REFERENCE BOXES
    # ========================================================

    label_path = labels_dir / f"{stem}.txt"

    gt_boxes = []

    if label_path.exists():

        for line in label_path.read_text().splitlines():

            parts = line.strip().split()

            if len(parts) != 5:
                continue

            cls, xc, yc, w, h = map(float, parts)

            if int(cls) != 0:
                continue

            xc *= img_w
            yc *= img_h
            w *= img_w
            h *= img_h

            gt_boxes.append([
                xc - w / 2,
                yc - h / 2,
                xc + w / 2,
                yc + h / 2
            ])


    gt_boxes = torch.tensor(
        gt_boxes,
        dtype=torch.float32
    ).reshape(-1, 4)


    # ========================================================
    # PREDICTIONS AT CONF >= 0.25
    # ========================================================

    if len(r.boxes) > 0:

        keep = (
            r.boxes.conf.cpu() >= CONF_THR
        )

        pred_boxes = (
            r.boxes.xyxy.cpu()[keep].float()
        )

    else:

        pred_boxes = torch.empty(
            (0, 4),
            dtype=torch.float32
        )


    n_gt = len(gt_boxes)
    n_pred = len(pred_boxes)


    # ========================================================
    # GREEDY ONE-TO-ONE MATCHING
    # ========================================================

    tp = 0

    if n_gt > 0 and n_pred > 0:

        ious = box_iou(
            gt_boxes,
            pred_boxes
        )

        candidates = []

        for g in range(n_gt):
            for p in range(n_pred):

                iou = float(ious[g, p])

                if iou >= IOU_THR:

                    candidates.append(
                        (iou, g, p)
                    )


        candidates.sort(
            reverse=True,
            key=lambda x: x[0]
        )

        used_gt = set()
        used_pred = set()

        for iou, g, p in candidates:

            if (
                g not in used_gt
                and
                p not in used_pred
            ):

                tp += 1

                used_gt.add(g)
                used_pred.add(p)


    fp = n_pred - tp
    fn = n_gt - tp


    total_TP += tp
    total_FP += fp
    total_FN += fn


    # ========================================================
    # IMAGE-LEVEL POSITIVE / NEGATIVE
    # ========================================================

    gt_positive = n_gt > 0
    pred_positive = n_pred > 0


    if gt_positive and pred_positive:
        img_TP += 1

    elif not gt_positive and pred_positive:
        img_FP += 1

    elif not gt_positive and not pred_positive:
        img_TN += 1

    elif gt_positive and not pred_positive:
        img_FN += 1


    rows.append({
        "image": stem,
        "GT_lesions": n_gt,
        "Predictions_conf025": n_pred,
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "GT_image_positive": gt_positive,
        "Pred_image_positive": pred_positive
    })


# ============================================================
# LESION-LEVEL METRICS
# ============================================================

lesion_precision = (
    total_TP /
    (total_TP + total_FP)
    if (total_TP + total_FP) > 0
    else 0
)

lesion_recall = (
    total_TP /
    (total_TP + total_FN)
    if (total_TP + total_FN) > 0
    else 0
)

lesion_f1 = (
    2 * lesion_precision * lesion_recall /
    (lesion_precision + lesion_recall)
    if (lesion_precision + lesion_recall) > 0
    else 0
)


# ============================================================
# IMAGE-LEVEL METRICS
# ============================================================

sensitivity = (
    img_TP /
    (img_TP + img_FN)
    if (img_TP + img_FN) > 0
    else 0
)

specificity = (
    img_TN /
    (img_TN + img_FP)
    if (img_TN + img_FP) > 0
    else 0
)

image_precision = (
    img_TP /
    (img_TP + img_FP)
    if (img_TP + img_FP) > 0
    else 0
)

accuracy = (
    (img_TP + img_TN) /
    66
)

image_f1 = (
    2 * image_precision * sensitivity /
    (image_precision + sensitivity)
    if (image_precision + sensitivity) > 0
    else 0
)


# ============================================================
# SAVE
# ============================================================

df = pd.DataFrame(rows)

save_path = (
    EXTERNAL_DIR /
    "MOPG7_External_FixedThreshold_ErrorAnalysis.csv"
)

df.to_csv(
    save_path,
    index=False
)


# ============================================================
# PRINT
# ============================================================

print("========================================")
print("LESION-LEVEL EXTERNAL ANALYSIS")
print("conf = 0.25 | IoU = 0.50")
print("========================================")

print("TP:", total_TP)
print("FP:", total_FP)
print("FN:", total_FN)

print("Precision:", round(lesion_precision, 4))
print("Recall:", round(lesion_recall, 4))
print("F1:", round(lesion_f1, 4))


print()
print("========================================")
print("IMAGE-LEVEL EXTERNAL ANALYSIS")
print("========================================")

print("TP:", img_TP)
print("FP:", img_FP)
print("TN:", img_TN)
print("FN:", img_FN)

print("Sensitivity:", round(sensitivity, 4))
print("Specificity:", round(specificity, 4))
print("Precision:", round(image_precision, 4))
print("Accuracy:", round(accuracy, 4))
print("F1:", round(image_f1, 4))

print()
print("Saved:", save_path)
```

### Step 291: Estimate bootstrap confidence intervals

Resamples complete images, rather than individual lesions, to preserve within-image dependence while estimating uncertainty.


```python
import numpy as np
import pandas as pd
from pathlib import Path

EXTERNAL_DIR = Path(
    "/content/drive/MyDrive/Pediatric_Caries_AI/external_validation/MOPG7_66"
)

csv_path = (
    EXTERNAL_DIR /
    "MOPG7_External_FixedThreshold_ErrorAnalysis.csv"
)

df = pd.read_csv(csv_path)

B = 2000
rng = np.random.default_rng(42)

results = []

for b in range(B):

    sample = df.sample(
        n=len(df),
        replace=True,
        random_state=int(rng.integers(0, 1_000_000_000))
    )

    gt = sample["GT_image_positive"].astype(bool)
    pred = sample["Pred_image_positive"].astype(bool)

    TP = int((gt & pred).sum())
    FP = int((~gt & pred).sum())
    TN = int((~gt & ~pred).sum())
    FN = int((gt & ~pred).sum())

    sensitivity = (
        TP / (TP + FN)
        if (TP + FN) > 0
        else np.nan
    )

    specificity = (
        TN / (TN + FP)
        if (TN + FP) > 0
        else np.nan
    )

    precision = (
        TP / (TP + FP)
        if (TP + FP) > 0
        else np.nan
    )

    accuracy = (
        (TP + TN) /
        (TP + FP + TN + FN)
    )

    f1 = (
        2 * precision * sensitivity /
        (precision + sensitivity)
        if (
            not np.isnan(precision)
            and not np.isnan(sensitivity)
            and (precision + sensitivity) > 0
        )
        else np.nan
    )

    results.append([
        sensitivity,
        specificity,
        precision,
        accuracy,
        f1
    ])


boot = pd.DataFrame(
    results,
    columns=[
        "Sensitivity",
        "Specificity",
        "Precision",
        "Accuracy",
        "F1"
    ]
)


gt_all = df["GT_image_positive"].astype(bool)
pred_all = df["Pred_image_positive"].astype(bool)
point_tp = int((gt_all & pred_all).sum())
point_fp = int((~gt_all & pred_all).sum())
point_tn = int((~gt_all & ~pred_all).sum())
point_fn = int((gt_all & ~pred_all).sum())
point_sensitivity = point_tp / (point_tp + point_fn) if (point_tp + point_fn) else 0.0
point_specificity = point_tn / (point_tn + point_fp) if (point_tn + point_fp) else 0.0
point_precision = point_tp / (point_tp + point_fp) if (point_tp + point_fp) else 0.0
point_accuracy = (point_tp + point_tn) / len(df) if len(df) else 0.0
point_f1 = (2 * point_precision * point_sensitivity / (point_precision + point_sensitivity)
            if (point_precision + point_sensitivity) else 0.0)
point_estimates = {
    "Sensitivity": point_sensitivity,
    "Specificity": point_specificity,
    "Precision": point_precision,
    "Accuracy": point_accuracy,
    "F1": point_f1,
}


print("========================================")
print("IMAGE-LEVEL BOOTSTRAP 95% CI")
print("========================================")

for metric in boot.columns:

    values = boot[metric].dropna()

    low = np.percentile(values, 2.5)
    high = np.percentile(values, 97.5)

    print(
        f"{metric:12s}: "
        f"{point_estimates[metric]:.4f} "
        f"(95% CI {low:.4f}–{high:.4f})"
    )


boot.to_csv(
    EXTERNAL_DIR /
    "MOPG7_ImageLevel_Bootstrap_2000.csv",
    index=False
)
```

### Step 292: Generate model predictions

Defines reusable logic used by the following preparation, evaluation, or visualization cells.


```python
PROJECT_DIR = Path("/content/drive/MyDrive/Pediatric_Caries_AI")
INTERNAL_IMAGES = PROJECT_DIR / "processed_data" / "yolo_caries_baseline" / "images" / "test"
INTERNAL_LABELS = PROJECT_DIR / "processed_data" / "yolo_caries_baseline" / "labels" / "test"
EXTERNAL_ROOT = PROJECT_DIR / "external_validation" / "MOPG7_66"
EXTERNAL_IMAGES = EXTERNAL_ROOT / "images" / "test"
EXTERNAL_LABELS = EXTERNAL_ROOT / "labels" / "test"
FIG_DIR = PROJECT_DIR / "final_model" / "figures"
FIG_DIR.mkdir(parents=True, exist_ok=True)

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image


def find_image_by_stem(folder, stem):
    exts = [".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"]

    for ext in exts:
        p = folder / f"{stem}{ext}"
        if p.exists():
            return p

    matches = list(folder.glob(stem + ".*"))

    if len(matches):
        return matches[0]

    raise FileNotFoundError(f"Image not found: {stem}")


def load_yolo_gt(label_path, width, height):
    """
    YOLO:
    class x_center y_center width height
    normalized
    """

    boxes = []

    if not label_path.exists():
        return np.empty((0, 4))

    text = label_path.read_text().strip()

    if not text:
        return np.empty((0, 4))

    for line in text.splitlines():
        parts = line.split()

        if len(parts) < 5:
            continue

        cls, xc, yc, bw, bh = map(float, parts[:5])

        xc *= width
        yc *= height
        bw *= width
        bh *= height

        x1 = xc - bw / 2
        y1 = yc - bh / 2
        x2 = xc + bw / 2
        y2 = yc + bh / 2

        boxes.append([x1, y1, x2, y2])

    return np.asarray(boxes, dtype=float)


def box_iou(a, b):

    x1 = max(a[0], b[0])
    y1 = max(a[1], b[1])
    x2 = min(a[2], b[2])
    y2 = min(a[3], b[3])

    inter = max(0, x2 - x1) * max(0, y2 - y1)

    area_a = max(0, a[2] - a[0]) * max(0, a[3] - a[1])
    area_b = max(0, b[2] - b[0]) * max(0, b[3] - b[1])

    union = area_a + area_b - inter

    if union <= 0:
        return 0.0

    return inter / union


def greedy_match(gt_boxes, pred_boxes, iou_thr=0.50):

    candidates = []

    for gi, g in enumerate(gt_boxes):
        for pi, p in enumerate(pred_boxes):
            iou = box_iou(g, p)

            if iou >= iou_thr:
                candidates.append((iou, gi, pi))

    candidates.sort(reverse=True)

    matched_gt = set()
    matched_pred = set()
    pairs = []

    for iou, gi, pi in candidates:

        if gi in matched_gt or pi in matched_pred:
            continue

        matched_gt.add(gi)
        matched_pred.add(pi)

        pairs.append((gi, pi, iou))

    unmatched_gt = [
        i for i in range(len(gt_boxes))
        if i not in matched_gt
    ]

    unmatched_pred = [
        i for i in range(len(pred_boxes))
        if i not in matched_pred
    ]

    return pairs, unmatched_gt, unmatched_pred


def run_prediction(image_path, confidence=0.25):

    r = model.predict(
        source=str(image_path),
        imgsz=1024,
        conf=confidence,
        iou=0.70,
        max_det=300,
        verbose=False
    )[0]

    if r.boxes is None or len(r.boxes) == 0:
        return (
            np.empty((0, 4)),
            np.empty((0,))
        )

    boxes = r.boxes.xyxy.cpu().numpy()
    confs = r.boxes.conf.cpu().numpy()

    return boxes, confs


def analyze_image(image_path, label_dir, confidence=0.25):

    image = Image.open(image_path).convert("RGB")

    width, height = image.size

    label_path = label_dir / f"{image_path.stem}.txt"

    gt = load_yolo_gt(
        label_path,
        width,
        height
    )

    pred, confs = run_prediction(
        image_path,
        confidence=confidence
    )

    pairs, unmatched_gt, unmatched_pred = greedy_match(
        gt,
        pred,
        iou_thr=0.50
    )

    best_iou = 0.0

    if len(gt) and len(pred):

        for g in gt:
            for p in pred:
                best_iou = max(
                    best_iou,
                    box_iou(g, p)
                )

    return {
        "image_path": image_path,
        "image": np.asarray(image),
        "gt": gt,
        "pred": pred,
        "confs": confs,
        "pairs": pairs,
        "unmatched_gt": unmatched_gt,
        "unmatched_pred": unmatched_pred,
        "best_iou": best_iou,
        "tp": len(pairs),
        "fp": len(unmatched_pred),
        "fn": len(unmatched_gt),
    }


def draw_panel(ax, record, panel_letter, subtitle):

    image = record["image"]

    ax.imshow(image)

    gt = record["gt"]
    pred = record["pred"]
    confs = record["confs"]

    matched_gt = {x[0] for x in record["pairs"]}
    matched_pred = {x[1] for x in record["pairs"]}

    # Matched reference annotation: RED
    for gi in matched_gt:

        x1, y1, x2, y2 = gt[gi]

        ax.add_patch(
            Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                edgecolor="red",
                linewidth=1.8
            )
        )

    # True-positive prediction: GREEN
    for pi in matched_pred:

        x1, y1, x2, y2 = pred[pi]

        ax.add_patch(
            Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                edgecolor="lime",
                linewidth=1.8
            )
        )

        ax.text(
            x1,
            max(0, y1 - 8),
            f"{confs[pi]:.2f}",
            fontsize=7,
            color="lime",
            bbox=dict(
                facecolor="black",
                alpha=0.55,
                pad=1
            )
        )

    # False negatives: BLUE dashed
    for gi in record["unmatched_gt"]:

        x1, y1, x2, y2 = gt[gi]

        ax.add_patch(
            Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                edgecolor="deepskyblue",
                linewidth=2,
                linestyle="--"
            )
        )

    # False positives: ORANGE
    for pi in record["unmatched_pred"]:

        x1, y1, x2, y2 = pred[pi]

        ax.add_patch(
            Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                edgecolor="orange",
                linewidth=1.7
            )
        )

        ax.text(
            x1,
            max(0, y1 - 8),
            f"{confs[pi]:.2f}",
            fontsize=7,
            color="orange",
            bbox=dict(
                facecolor="black",
                alpha=0.55,
                pad=1
            )
        )

    ax.set_title(
        f"({panel_letter}) {subtitle}\n"
        f"TP={record['tp']}, FP={record['fp']}, FN={record['fn']}",
        fontsize=10
    )

    ax.axis("off")
```

### Step 295: A: best near-localization

Performs the next reproducible operation in this section and exposes its intermediate result for inspection.


```python
positive = [
    r for r in external_records
    if len(r["gt"]) > 0
]

negative = [
    r for r in external_records
    if len(r["gt"]) == 0
]

# A: best near-localization
near_case = max(
    positive,
    key=lambda r: r["best_iou"]
)

# B: strongest missed positive case
positive_no_pred = [
    r for r in positive
    if len(r["pred"]) == 0
]

if positive_no_pred:
    missed_case = max(
        positive_no_pred,
        key=lambda r: len(r["gt"])
    )
else:
    missed_case = max(
        positive,
        key=lambda r: r["fn"]
    )

# C: negative image with most false positives
negative_fp_case = max(
    negative,
    key=lambda r: r["fp"]
)

# D: positive image with many false positives
remaining = [
    r for r in positive
    if r["stem"] not in {
        near_case["stem"],
        missed_case["stem"]
    }
]

positive_fp_case = max(
    remaining,
    key=lambda r: (r["fp"], r["fn"])
)

selected = [
    (
        near_case,
        f"Near-localization; max IoU={near_case['best_iou']:.3f}"
    ),
    (
        missed_case,
        "Caries-positive image with missed lesion(s)"
    ),
    (
        negative_fp_case,
        "Caries-negative image with false positives"
    ),
    (
        positive_fp_case,
        "Positive image with localization errors"
    )
]

print("\nSelected cases:")

for rec, title in selected:
    print(
        rec["stem"],
        "|", title,
        "| GT=", len(rec["gt"]),
        "| Pred=", len(rec["pred"]),
        "| TP=", rec["tp"],
        "| FP=", rec["fp"],
        "| FN=", rec["fn"],
        "| Best IoU=", round(rec["best_iou"], 4)
    )
```
